# Practical 4: Inventory Management System using Lists and Dictionaries

**Problem Statement** 

Develop an inventory management system for a small stationery shop. Each product should contain Product ID, Product Name, Quantity Available, and Price. The system should allow users to manage inventory records. 

**Functional Requirements** 

The program should provide a menu-driven interface to: 

        ● Add a new product 
        ● Search a product using Product ID 
        ● Update product quantity 
        ● Delete a product 
        ● Display all products 
        ● Calculate total inventory value 
        
**Inventory Value = Quantity × Price**

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Product Name" : [],
    "Product ID" : [],
    "Quantity" : [],
    "Price" : [],
    "Inventory Values" : []
}

df = pd.DataFrame(data)

df['Product Name'] = df['Product Name'].astype(str)
df['Product ID'] = df['Product ID'].astype(str)
df['Quantity'] = df['Quantity'].astype(int)
df['Price'] = df['Price'].astype(float)
df['Inventory Values']= df['Inventory Values'].astype(float)

df.info()
```

```python
def add_product():
    global df
    pid = input("Enter Product ID: ")
    
    if pid in df["Product ID"].values:
        print("Warning! Product ID already exists! Please use a unique ID.\n")
        return
    
    pname = input("Enter Product Name: ")
    qty = int(input("Enter Quantity: "))
    price = float(input("Enter Price: "))
    inv_value = qty * price
    
    new_row = {
        "Product Name": pname,
        "Product ID": pid,
        "Quantity": qty,
        "Price": price,
        "Inventory Values": inv_value
    }
    df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
    print("Product added successfully!\n")

def search_product():
    pid = input("Enter Product ID to search: ")
    result = df[df["Product ID"] == pid]
    if not result.empty:
        print("Product Found:\n", result, "\n")
    else:
        print("Product not found!\n")

def update_quantity():
    global df
    pid = input("Enter Product ID to update quantity: ")
    if pid in df["Product ID"].values:
        new_qty = int(input("Enter new quantity: "))
        df.loc[df["Product ID"] == pid, "Quantity"] = new_qty
        df.loc[df["Product ID"] == pid, "Inventory Values"] = (
            df.loc[df["Product ID"] == pid, "Quantity"] *
            df.loc[df["Product ID"] == pid, "Price"]
        )
        print("Quantity updated successfully!\n")
    else:
        print("Product not found!\n")

def delete_product():
    global df
    pid = input("Enter Product ID to delete: ")
    if pid in df["Product ID"].values:
        df = df[df["Product ID"] != pid]
        print("Product deleted successfully!\n")
        df.reset_index(drop=True,inplace=True)
    else:
        print("Product not found!\n")

def display_products():
    if df.empty:
        print("Inventory is empty!\n")
    else:
        print("\nInventory Records:\n", df, "\n")

def calculate_value():
    total_value = df["Inventory Values"].sum()
    print(f"Total Inventory Value: {total_value}\n")

def menu():
    print("====== Stationery Shop Inventory Management ======")
    print("1. Add New Product")        
    print("2. Search Product by ID")
    print("3. Update Product Quantity")
    print("4. Delete Product")
    print("5. Display All Products")
    print("6. Calculate Total Inventory Value")
    
    choice = input("Enter your choice (1-7): ")
        
    if choice == "1":
        add_product()
    elif choice == "2":
        search_product()
    elif choice == "3":
        update_quantity()
    elif choice == "4":
        delete_product()
    elif choice == "5":
        display_products()
    elif choice == "6":
        calculate_value()
    else:
        print("Invalid choice! Please try again.\n")
```

```python
menu()
```
