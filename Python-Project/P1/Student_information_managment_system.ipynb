**Practical 1: Student Information Management System** 



**Problem Statement :**

 Develop a Python program that collects the details of a student, including Enrollment Number, Name, Branch, Semester, Date of Birth, Mobile Number, and CGPA. The program should display all information in a neatly formatted report similar to a college student profile. 

Functional Requirements
    
        ● Accept all details from the user.
        ● Store data using appropriate variables. 
        ● Display the information using formatted output. 
        ● Calculate whether the student is eligible for placement based on CGPA (for example, CGPA ≥ 6.5). 

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Enrollment No" : [],
    "Name" : [],
    "Branch" : [],
    "Semester" : [],
    "Date Of Birth" : [],
    "Mobile Number" : [],
    "CGPA" : []
}

df = pd.DataFrame(data)

df["Enrollment No"] = df["Enrollment No"].astype("Int64")      
df["Name"] = df["Name"].astype("string")                         
df["Branch"] = df["Branch"].astype("string")                     
df["Semester"] = df["Semester"].astype("Int64")                  
df["Date Of Birth"] = pd.to_datetime(df["Date Of Birth"])        
df["Mobile Number"] = df["Mobile Number"].astype("Int64")        
df["CGPA"] = df["CGPA"].astype("float64")
```

```python
ALLOWED_BRANCHES = ["CSE", "IT", "EC", "ME", "EE", "AIML", "IC", "CE"]

def add_data(en : int):
    global df
    # check Duplicate Enrollment no 
    if en in df["Enrollment No"].values:
        print(f"Error: Enrollment No {en} already exists!")
        return df
    else:
        name = input("Enter the name : ")
    
        # Branch validation
        branch = input(f"Enter the Branch name {ALLOWED_BRANCHES}: ")
        if branch not in ALLOWED_BRANCHES:
            print("Error: Invalid branch!")
            return df
        
        # Semester validation
        sem = int(input("Enter your current Semester (1-8): "))
        if sem < 1 or sem > 8:
            print("Error: Semester must be between 1 and 8!")
            return df
        
        # DOB validation
        dob = pd.to_datetime(input("Enter your Date Of Birth (YYYY-MM-DD): "), errors="coerce")
        if pd.isna(dob):
            print("Error: Invalid date format!")
            return df
        
        # Mobile number validation
        mn = input("Enter your Mobile Number (10 digits): ")
        if not mn.isdigit() or len(mn) != 10:
            print("Error: Mobile number must be exactly 10 digits!")
            return df
        mn = int(mn)
        
        # CGPA validation
        cgpa = float(input("Enter your CGPA (0–10, two decimals): "))
        if cgpa < 0 or cgpa > 10:
            print("Error: CGPA must be between 0 and 10!")
            return df
        cgpa = round(cgpa, 2)  # only two fraction digits

    new_row = {
        "Enrollment No" : en,
        "Name" : name,
        "Branch" : branch,
        "Semester" : sem,
        "Date Of Birth" : dob,
        "Mobile Number" : mn,
        "CGPA" : cgpa
    }

    df = pd.concat([df,pd.DataFrame([new_row])],ignore_index=True)
    return df
```

```python
ch = input("IF you want to add details then prees 1  and get details then press 2 and if you want to check eligiblity then press 3")

match(ch) :
    case '1' :
        en = int(input("Enter the Enrollment No : "))
        add_data(en)
        display(df)
    case '2':
        display(df)
    case '3':
        ch_cgpa = int(input("Enter the student enrollment no : "))
        if ch_cgpa in df["Enrollment No"].values:
            student_row = df.loc[df['Enrollment No'] == ch_cgpa]
            if student_row["CGPA"].values[0] >=6.5:
                print("Student is Eligible for the placment")
            else:
                print("Student not eligible for the placment")
        
```
