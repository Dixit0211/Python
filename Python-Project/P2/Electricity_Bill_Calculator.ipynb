# Practical 2: Electricity Bill Calculator 

**Problem Statement :**

 Develop a Python program that calculates the monthly electricity bill based on the number of units consumed. Different tariff rates should be applied according to predefined slabs. The final bill should also include fixed charges and GST. 

**Sample Tariff**

 Units      Rate (₹/Unit)
 
 0–100  :        3.50

 101–300  :      5.00

 Above 300  :    7.50 
 
 Fixed Charge = ₹100 

 GST = 5% 
 
 **Functional Requirements**

    ● Accept customer name and units consumed. 
    ● Calculate bill according to slab rates. 
    ● Add fixed charges and GST. 
    ● Display a professional electricity bill. 

```python
import pandas as pd  
import numpy as np
```

```python
data = {
    "Customer Name" : [],
    "Units Consume" : [],
    "Total Units Prizes" : [],
    "GST Bill" : [],
    "Total Bill" : []
}

df = pd.DataFrame(data)
df['Customer Name'] = df['Customer Name'].astype(str)
df['Units Consume'] = df['Units Consume'].astype('Int64')
df['Total Units Prizes'] = df['Total Units Prizes'].astype('Float64')
df['GST Bill'] = df['GST Bill'].astype('Float64')
df['Total Bill'] = df['Total Bill'].astype('Float64')

df.info()
```

```python
def total_units_prizes(unit):
    prize  = 1
    if unit in range(0,101) :
        prize = unit *  3.50
        return prize
    elif unit in range(101,201):
        prize = unit * 5.00
        return prize
    else :
        prize = unit * 7.50
        return prize


def gst_bill(unit):
    unit_prize = total_units_prizes(unit)
    prize = unit_prize * 0.05
    return prize

def total_Bill(unit):
    print("100 is fixed charge")
    unit_prize = total_units_prizes(unit)
    gst_prize = gst_bill(unit)
    return (100 + unit_prize + gst_prize)

def add_details():
    global df
    print("Enter the details.......")
    name = input("Enter the consumer name : ")
    unit_consume = int(input("Enter the total number of unit consume :"))
    totalUnitBill = total_units_prizes(unit_consume)
    totalGstBill = gst_bill(unit_consume)
    totalBill = total_Bill(unit_consume)

    row_data = {
        "Customer Name" : name,
        "Units Consume" : unit_consume,
        "Total Units Prizes" : totalUnitBill,
        "GST Bill" : totalGstBill,
        "Total Bill" : totalBill
    }

    df = pd.concat([df,pd.DataFrame([row_data])],ignore_index=True)
    return df
```

```python
ch = input(" if you want to add details press then '1' or if you want to see details then press '2' : ")
match(ch):
    case '1':
        add_details()
    case '2':
        display(df)
```
