# Practical 3: Student Grade Analyzer 

**Problem Statement**

 Develop a program that accepts marks obtained by a student in five subjects. The program should calculate the total marks, percentage, grade, and pass/fail status. Display the complete result in a formatted report card. 

**Grade Policy(Example)**

Percentage      Grade

≥90         :      A+

80–89       :      A

70–79        :     B

60–69         :    C

50–59          :   D

Below 50      :   Fail 



Functional Requirements 

        ● Accept marks of five subjects. 
        ● Validate that marks are between 0 and 100. 
        ● Calculate total and percentage. 
        ● Assign grade. 
        ● Display pass/fail status. 
        ● Identify the highest and lowest scoring subjects.

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Enrollment No" : [],
    "Name" : [],
    "Artificial Intelligence" : [],
    "Machine Learning" : [],
    "Digital Image Processing" : [],
    "Data Mining Techniques" : [],
    "Python Programming" : [],
    "Total Marks" : [],
    "Percentage" : [],
    "Grade" : [],
    "Status" : []
}

df = pd.DataFrame(data)

df['Enrollment No'] = df["Enrollment No"].astype("Int64")
df['Name'] = df["Name"].astype('str')
df['Artificial Intelligence'] = df["Artificial Intelligence"].astype('int64')
df['Machine Learning'] = df["Machine Learning"].astype('int64')
df['Digital Image Processing'] = df["Digital Image Processing"].astype('int64')
df['Data Mining Techniques'] = df['Data Mining Techniques'].astype('int64')
df['Python Programming'] = df["Python Programming"].astype('int64')
df['Total Marks'] = df["Total Marks"].astype('int64')
df['Percentage'] = df['Percentage'].astype('float64')
df['Grade'] = df['Grade'].astype('str')
df['Status'] = df['Status'].astype('str')

df.info()
```

```python
def validation():
    mark = int(input())
    if mark in range(0, 101):
        return mark
    else:
        print("You entered wrong details (marks between 0-100)")
        return validation()   # ask again until valid

def percentage(total_marks):
    return (total_marks / 500) * 100

def grade(per):
    if per >= 90:
        return 'A+'
    elif per >= 80:
        return 'A'
    elif per >= 70:
        return 'B'
    elif per >= 60:
        return 'C'
    elif per >= 50:
        return 'D'
    else:
        return 'Fail'

def status(gd):
    return 'Fail' if gd == 'Fail' else 'Pass'

def add_init():
    en = int(input("Enter the Enrollment No : "))
    global df
    if en in df['Enrollment No'].values:
        print("Warning! You already added details")
    else:
        name = input("Enter the name : ")
        print("Enter the marks of Artificial Intelligence : ")
        ai = validation()
        print("Enter the marks of Machine Learning : ")
        ml = validation()
        print("Enter the marks of Digital Image Processing : ")
        dip = validation()
        print("Enter the marks of Data Mining Techniques : ")
        dmt = validation()
        print("Enter the marks of Python Programming : ")
        pp = validation()

        total_marks = ai + ml + dip + dmt + pp
        per = percentage(total_marks)
        gr = grade(per)
        st = status(gr)

        row_data = {
            "Enrollment No": [en],
            "Name": [name],
            "Artificial Intelligence": [ai],
            "Machine Learning": [ml],
            "Digital Image Processing": [dip],
            "Data Mining Techniques": [dmt],
            "Python Programming": [pp],
            "Total Marks": [total_marks],
            "Percentage": [per],
            "Grade": [gr],
            "Status": [st]
        }

        df = pd.concat([df, pd.DataFrame(row_data)], ignore_index=True)
        print("Student details added successfully!")
```

```python
def find_highest_lowest_by_enrollment(enrollment_no):
    global df
    if enrollment_no not in df['Enrollment No'].values:
        print("Enrollment number not found!")
        return
    
    student_row = df.loc[df['Enrollment No'] == enrollment_no].iloc[0]

    marks_dict = {
        "Artificial Intelligence": student_row["Artificial Intelligence"],
        "Machine Learning": student_row["Machine Learning"],
        "Digital Image Processing": student_row["Digital Image Processing"],
        "Data Mining Techniques": student_row["Data Mining Techniques"],
        "Python Programming": student_row["Python Programming"]
    }

    highest_subject = max(marks_dict, key=marks_dict.get)
    lowest_subject = min(marks_dict, key=marks_dict.get)

    print(f"Enrollment No: {enrollment_no}")
    print(f"Highest Scoring Subject: {highest_subject} ({marks_dict[highest_subject]} marks)")
    print(f"Lowest Scoring Subject: {lowest_subject} ({marks_dict[lowest_subject]} marks)")
```

```python
ch = input("IF you want to add details then press 1 \n if you want to see details 2 \n if you want to see highest and lowest subject then press 3 ")

match(ch):
    case '1':
        add_init()
        display(df)
    case '2':
        display(df)
    case '3':
        en = int(input("Enter the enrollment no :"))
        find_highest_lowest_by_enrollment(en)
```
