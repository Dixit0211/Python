# Practical 6: Password Strength Validator 

**Problem Statement** 

Develop a Python program that checks whether a password satisfies basic cybersecurity requirements before allowing a user to register. 

**Password Policy** 

A valid password must contain:

    ● Minimum 8 characters 
    ● At least one uppercase letter 
    ● At least one lowercase letter 
    ● At least one digit 
    ● At least one special character (@, #, $, %, &, *) 


**Functional Requirements** 

    ● Accept password from the user. 
    ● Check each validation rule. 
    ● Display specific messages indicating which criteria are not satisfied. 
    ● Display "Strong Password" only if all conditions are met. 

```python

# Check password minimum length
def check_length(password):
    return len(password) >= 8

# Check uppercase
def check_uppercase(password):
    return any(ch.isupper() for ch in password)

# Check lowercase
def check_lowercase(password):
    return any(ch.islower() for ch in password)

# Check digit
def check_digit(password):
    return any(ch.isdigit() for ch in password)

# Check special character
def check_special(password):
    return any(ch in "@#$%&*" for ch in password)

# Validate password strength
def validate_password(password):
    missing = []

    if not check_length(password):
        missing.append("✓ Minimum 8 Characters")
    if not check_uppercase(password):
        missing.append("✓ Uppercase Letter")
    if not check_lowercase(password):
        missing.append("✓ Lowercase Letter")
    if not check_digit(password):
        missing.append("✓ Digit")
    if not check_special(password):
        missing.append("✓ Special Character (@, #, $, %, &, *)")

    if not missing:
        print("Password Strength: Strong ")
    else:
        print("Missing:")
        for rule in missing:
            print(rule)
        print("Password Strength: Weak")

password = input("Enter Password: ")
validate_password(password)
```
