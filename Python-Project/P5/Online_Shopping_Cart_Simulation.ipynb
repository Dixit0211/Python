# Practical 5: Online Shopping Cart Simulation 

**Problem Statement** 

Develop a shopping cart application similar to an e-commerce website. Users should be able to browse available products, add items to their cart, remove items, and generate the final invoice. 

**Functional Requirements** 

    ● Display available products. 
    ● Add products to the cart. 
    ● Remove products from the cart. 
    ● Modify quantities. 
    ● Calculate subtotal. 
    ● Apply discount: 
        ○ 10% discount for purchases above ₹5000. 
    ● Calculate GST (18%). 
    ● Generate the final bill. 

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Product ID" : [],
    "Product Name" : [], 
    "Quantity" : [], 
    "Price" : [],
    "Total" : []
}

df = pd.DataFrame(data)

df['Product ID'] = df["Product ID"].astype(str)
df['Product Name'] = df['Product Name'].astype(str)
df['Quantity'] = df['Quantity'].astype(int)
df['Price'] = df['Price'].astype(float)
df['Total'] = df['Total'].astype(float)

df.info()
```

```python
row_data = {
    "Product ID": ["P001", "P002", "P003", "P004", "P005", "P006", "P007", "P008"],
    "Product Name": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones", "Smartphone", "USB Cable", "External HDD"],
    "Quantity": [1, 2, 1, 1, 1, 1, 3, 1],
    "Price": [45000, 1200, 2500, 15000, 3500, 30000, 300, 6000],
    "Total": [45000, 2400, 2500, 15000, 3500, 30000, 900, 6000]
}

df1 = pd.DataFrame(row_data)
df = pd.concat([df,df1],ignore_index=True)
df
```

```python
def search_product():
    global df
    pid = input("Enter Product ID to search: ")
    if pid in df["Product ID"].values:
        product = df[df["Product ID"] == pid].iloc[0]
        print("\n Product Found")
        print("---------------------------")
        print(f"Product ID   : {product['Product ID']}")
        print(f"Product Name : {product['Product Name']}")
        print(f"Quantity     : {product['Quantity']}")
        print(f"Price        : ₹{product['Price']}")
        print(f"Total        : ₹{product['Total']}")
        print("---------------------------\n")
    else:
        print(" Product not found!\n")

def add_to_cart():
    global df
    pid = input("Enter Product ID to add: ")
    pname = input("Enter Product Name: ")
    qty = int(input("Enter Quantity: "))
    price = float(input("Enter Price: "))
    total = qty * price

    if pid in df["Product ID"].values:
        df.loc[df["Product ID"] == pid, "Quantity"] += qty
        df.loc[df["Product ID"] == pid, "Total"] = df.loc[df["Product ID"] == pid, "Quantity"] * df.loc[df["Product ID"] == pid, "Price"]
        print(" Updated existing product in cart!\n")
    else:
        new_product = {
            "Product ID": [pid],
            "Product Name": [pname],
            "Quantity": [qty],
            "Price": [price],
            "Total": [total]
        }
        new_df = pd.DataFrame(new_product)
        df = pd.concat([df, new_df], ignore_index=True)
        print(" Product added to cart!\n")

def modify_quantity():
    global df
    pid = input("Enter Product ID to modify: ")
    if pid in df["Product ID"].values:
        new_qty = int(input("Enter new quantity: "))
        if new_qty > 0:
            df.loc[df["Product ID"] == pid, "Quantity"] = new_qty
            df.loc[df["Product ID"] == pid, "Total"] = new_qty * df.loc[df["Product ID"] == pid, "Price"]
            print(" Quantity updated!\n")
        else:
            df = df[df["Product ID"] != pid]
            print(" Product removed (quantity set to 0)!\n")
    else:
        print(" Product not found in cart!\n")

def apply_discount(subtotal):
    discount = 0
    if subtotal > 5000:
        discount = 0.10 * subtotal
    return discount

def calculate_gst(amount_after_discount):
    gst = 0.18 * amount_after_discount
    return gst

def generate_final_bill(df):
    if df.empty:
        print(" Cart is empty!\n")
        return
    
    print("\n Final Invoice")
    print("Product          Qty   Price       Total")
    print("----------------------------------------")
    
    for _, row in df.iterrows():
        print(f"{row['Product Name']:<15} {row['Quantity']:>3}   ₹{row['Price']:>8,.2f}   ₹{row['Total']:>10,.2f}")
    
    print("----------------------------------------")
    
    subtotal = df["Total"].sum()
    print(f"Subtotal           ₹{subtotal:,.2f}")
    
    discount = apply_discount(subtotal)
    if discount > 0:
        print(f"Discount           ₹{discount:,.2f}")
    
    after_discount = subtotal - discount
    gst = calculate_gst(after_discount)
    print(f"GST                ₹{gst:,.2f}")
    print("----------------------------------------")
    final_total = after_discount + gst
    print(f"Total Amount       ₹{final_total:,.2f}\n")
```

```python
def main_menu():
    while True:
        print("\n--- Shopping Cart Menu ---")
        print("1. Add to Cart")
        print("2. Modify Quantity")
        print("3. Search Product")
        print("4. View Invoice")
        print("5. Exit")
        
        choice = input("Enter choice: ")
        
        if choice == "1":
            add_to_cart()
        elif choice == "2":
            modify_quantity()
        elif choice == "3":
            search_product()
        elif choice == "4":
            generate_final_bill(df)
        elif choice == "5":
            print("Thank you for shopping! 🛒")
            break
        else:
            print("Invalid choice, try again.")
```

```python
main_menu()
```
