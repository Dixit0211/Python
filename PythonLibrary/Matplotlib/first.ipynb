```python
"""
matplotlib is aploatting library for the python progrmming langauge and its numericals mathematics extensions numpy.
it provide an object-oriented api for embedding ploats into application using genrel-purpose GUI toolkits like Tkinter,
wxPython,Qt, or GTK+.

some of the major pros of Matplotlib are:
    generally easy to get started for simple plots
    supports for custom labels and texts
    great control of every element in sfigure
    high-quality output in many formula
    very customizable in genral
"""

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

## simple Example

x = np.arange(0,10)
y = np.arange(11,21)

##plotting using matplotlib 
    # plt scater
plt.scatter(x,y,c='r')  # c indicte the color
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Graph in 2D')
plt.savefig('Test.png') # to sam efigure in perticular png ang jpeg file
plt.show()

y = x**2

plt.plot(x,y,'r*--',linestyle = 'dashed', linewidth=2, markersize = 12) 
plt.show()
```

```python
x = np.arange(1,11)
y = 3*x + 5
plt.title('Matplotlib demo')
plt.xlabel('x axis caption')
plt.ylabel('y axis caption')
plt.plot(x,y)
plt.show()
```

```python
# we make wave form diagram

x = np.arange(0,4 * np.pi, 0.1) #(third perameter is step)
y = np.sin(x)

plt.title("Sine wave form")

#plot the points using maatplotlib
plt.plot(x,y)
plt.show()
```

```python
x = np.arange(0,5*np.pi,0.1)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.subplot(2,1,1)
plt.plot(x,y_sin)
plt.subplot(2,1,2)
plt.plot(x,y_cos)
plt.show()
```

```python
# bar plot

x = [2,8,10]
y = [11,16,9]

x2 = [3,9,11]
y2 = [6,15,7]

plt.bar(x,y, align='center')
plt.bar(x2,y2, color = 'g')
plt.title('Bar graph')
plt.ylabel('Y axis')
plt.xlabel('X axis')
plt.show()
```

```python
# pie chart

labels = 'python','C++','Ruby', 'Java'
sizes = [215,130,245,210]
colors = ['gold','yellowgreen', 'lightcoral','lightskyblue']
explode  = (0.1, 0, 0, 0) # explode 1st slice how far this goes

# plot
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True)
plt.show() # autopct is what formate data you want
```
